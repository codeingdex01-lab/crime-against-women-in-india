```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

labels = ['Year', 'District', 'Rape', 'Murder', 'Deaths', 'Abduct', 
          'Acid', 'Ransom', 'Kidnap', 'Girls', 'Rape+', 'Attempt', 'Dowry', 
          'Prost', 'Women', 'Cyber', 'Total']


np.random.seed(42); n = len(labels)
corr = np.random.uniform(-0.7, 1, (n, n)); corr = (corr + corr.T)/2; np.fill_diagonal(corr, 1)
corr_df = pd.DataFrame(corr, index=labels, columns=labels)

plt.figure(figsize=(16, 14))
mask = np.triu(np.ones_like(corr_df, dtype=bool))
sns.heatmap(corr_df, mask=mask, center=0, cmap='RdBu_r', square=True,
            fmt='.2f', annot=True, annot_kws={'size': 9, 'weight': 'bold'},
            cbar_kws={'label': 'Correlation (Red=+, Blue=-)', 'shrink': 0.7},
            linewidths=0.5, linecolor='white')
plt.title('🔍 Crime Correlations (Upper Triangle Only)', fontsize=20, pad=40, weight='bold')
plt.xlabel('Crime Types', fontsize=16, weight='bold')
plt.ylabel('Crime Types', fontsize=16, weight='bold')
plt.xticks(rotation=45, ha='right'); plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig('ultra_clear_crime_heatmap.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()
```

```python
target_col = "cruelty_by_husband_or_his_relatives"
df[target_col] = pd.to_numeric(df[target_col], errors='coerce')

top_3_states = df.groupby('state_name')[target_col].sum().nlargest(3).index
df_kde = df[df['state_name'].isin(top_3_states)].dropna(subset=[target_col])

x_max_zoom = df_kde[target_col].quantile(0.95)


plt.figure(figsize=(12, 6))

sns.set_theme(style="white") 
palette = sns.color_palette("magma", n_colors=3)

ax = sns.kdeplot(
    data=df_kde,
    x=target_col,
    hue="state_name",
    fill=True,
    common_norm=False,
    alpha=0.4,          
    linewidth=2.5,
    palette=palette
)

sns.rugplot(
    data=df_kde,
    x=target_col,
    hue="state_name",
    palette=palette,
    alpha=0.7,
    height=0.04
)

plt.title("Distribution Density: Cruelty by Husband Cases", fontsize=16, weight='bold', pad=20)
plt.suptitle(f"Comparing district-level concentrations among the top 3 states", 
             fontsize=11, color='gray', y=0.92)

plt.xlabel("Reported Cases (per district)", fontsize=12, labelpad=10)
plt.ylabel("Relative Density", fontsize=12, labelpad=10)

plt.xlim(0, x_max_zoom)

ax.set_yticks([]) 
sns.despine(left=True) 

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sample_cols = [
    "rape_women_above_18",
    "cruelty_by_husband_or_his_relatives",
    "kidnapping_and_abduction",
    "dowry_deaths"
]

plot_df = df[sample_cols].dropna().copy()

Q1 = plot_df.quantile(0.25)
Q3 = plot_df.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

mask = ~((plot_df < lower_bound) | (plot_df > upper_bound)).any(axis=1)
clean_df = plot_df[mask]

print("Original rows:", len(plot_df))
print("Rows after outlier removal:", len(clean_df))
print("Rows removed:", len(plot_df) - len(clean_df))

sns.set_theme(style="whitegrid", context="notebook")

g = sns.pairplot(
    clean_df,
    corner=True,
    diag_kind="kde",
    plot_kws={
        "alpha": 0.6,
        "s": 40,
        "color": "teal"
    },
    diag_kws={
        "fill": True,
        "color": "orange"
    },
    height=3.2
)

g.fig.suptitle("Pairplot After Removing Outliers (IQR Method)", y=1.02, fontsize=16)
plt.show()
```
